Lambda Functions (Theoretical Level)

A lambda function in Python is:
1/ An __anonymous__ function (it doesn't need any name)
(but it can be assigned to a variable, if needed)

2/ A small, single-expression function > just a single line
(so we don't have to define it with the syntax of normal functions)

3/ Syntax:
``` lambda arguments: expression```, where:

- arguments = something we pass into function (lambda takes any number of arguments, even zero),
- expression = what do we want it to do? what kind of output we expect?

__WHY?__ 
to calculate something "on the fly", short tasks, without defining it properly

````python
# sort lists
numbers = [5, 2, 9, 1]

# Ascending (default)
sorted_numbers = sorted(numbers)
print(sorted_numbers) 

# Descending using lambda
sorted_numbers_desc = sorted(numbers, key=lambda x: -x)
print(sorted_numbers_desc)
````

````python
# sort strings by length
words = ['apple', 'banana', 'kiwi']

sorted_words = sorted(words, key=lambda w: len(w))
print(sorted_words)
````

````python
# sort tuples by second element (age)
people = [('Alice', 25), ('Bob', 20), ('Charlie', 30)]

sorted_people = sorted(people, key=lambda person: person[1])
print(sorted_people)  
````

````python
# without lambda:
sorted_people = sorted(people)
print(sorted_people)  
````

````python
# sort dicts

data = [{'name': 'Alice', 'score': 90}, {'name': 'Bob', 'score': 75}]

# Sort by score
sorted_data = sorted(data, key=lambda d: d['score'], reverse=True)
print(sorted_data)
````

## Higher-Order Functions ## 
A higher-order function is a function that can do at least one of these:

1.	Take another function as an argument (we saw the example with ```sorted```)
2.	Return a function as a result (sounds like magic)

This concept comes from __functional programming__, and it makes code more flexible and expressive.

````python
# take a function as an argument:
def apply_function(func, value): # takes two arguments: a function and a value
    return func(value) # returns???

result = apply_function(lambda x: x**2, 5)
print(result)
````

````python
type(result) # why is it an integer? > we passed an integer
````

````python
result = apply_function(lambda x: x*18, 'hello ')
print(result) # we got a string
````

````python
# or shorter:
result = apply_function(lambda x: x**2, 5) # we define the function
print(result)
````

````python
# function returns a function (as an output)
def make_multiplier(n): # takes a variable as an argument
    return lambda x: x * n # returns a function

double = make_multiplier(2)
print(double(5)) 

type(double) # function
````

````python
# Task: Define a function that receives a list and a function as an argument
# this argument is a filter to apply to the list (to each element)
# Returns a new list based on the filter

def filter_me(list, func):
    for element in list:
        if func(element) == True:
            print(element)

````

````python
my_list = [1, 10, 20, 40]
filter_func = lambda x: x<20
filter_me(my_list,filter_func)
````
