````python
# Update the function definition based on the Pseudocode above
def list_of_elements(data, col_index_local):
    '''
    This function gets a list of unique values from a column.
    It takes TWO parameters:
    1/ data represents some data (e.g., list of strings)
    2/ col_index_local represent the index of particular column in the dataset

    Returns a list of unique values in a given column.
    '''

    result_list = [] # empty list for the results

    # checking the quality of the data:
    if len(data) == 0:
        print('There is an empty dataset, nothing to explore.')

    # checking the quality of index:
    if col_index_local > len(data[0]):
        print('Your index is not valid, it does not exist in the given dataset.')
        return None

    else: 

        for row in data: # looping over all rows in the data
            cell = row[col_index_local] # getting a particular cell in the row (with the index we want)

            if cell not in result_list: # checking if it's already in the list >> unique?
                result_list.append(cell) # if no, adding to the list
    
    return result_list # return the list with results based on given parameters
````

````python
def elements_count(my_data_local, col_index_local, sep_local=','):
    element_count_dict = {}
    """
  This function elements_count first creates an empty dictionary called element_count_dict which will 
store each unique element as a key and its count as value. Then it loops through each row i in the dataset my_data_local.
For each row, it checks if the value at column col_index_local is empty, and if it is empty, it just skips that row using continue. 
If the cell has data, it creates an empty list genres, then splits the string i[col_index_local] by the separator sep_local and for each part j, 
it removes spaces using strip() and appends it to genres. Next, it loops through each element gen in the list genres and checks
if gen is already in element_count_dict. If not, it adds gen as a key with value 1, otherwise it increases the value by 1. 
After counting all elements, it sorts element_count_dict by the counts using sorted with a lambda function, 
then converts it back to a dictionary. Finally, it returns element_count_dict which now has all unique elements and their counts in descending order.
    """
    for i in my_data_local:
        value = i[col_index_local]
        if not value:
            continue
        genres = []
        for j in value.split(sep_local):
            genres.append(j.strip())
        
        # Count each genre
        for gen in genres:
            if gen not in element_count_dict.keys():
                element_count_dict[gen] = 1
            else:
                element_count_dict[gen] += 1
    
    # Sorting dictionary by count in descending order
    
    element_count_dict = dict(
        sorted(element_count_dict.items(), key=lambda x: x[1], reverse=True)
    )
    
    return element_count_dict
````

````python
def duration_conversion(data, col_index_local=9):
    '''
    This function converts current format of movies and shows to a cleaner numeric version.
    For movies, it removes word "minutes" and just keeps the amount of minutes.
    For tv shows, it removes word "season" or "seasons" and just keeps the number of seasons.
    It doesn't  do anything for other formats.
    
    The function takes two arguments:
    1/ data = represents the dataset to iterate over
    2/ col_index_local = represents the number of a columnt to work with. 
    By default, it's 9 because "duration" column in the current dataset is 9. But this can be changed for other datasets.
    
    The function doesn't return anything but updates the value in column. -- should return something...
    '''

    # checking the quality of the data:
    if len(data) == 0:
        print('There is an empty dataset, nothing to explore.')

    # checking the quality of index:
    if col_index_local > len(data[0]):
        print('Your index is not valid, it does not exist in the given dataset.')

    for row in data:
        working_cell = row[col_index_local] # grab a cell in specific column in a row

        if not working_cell: # pass the empty ones
            continue 

        working_cell = str(working_cell).strip() # remove whitespaces on right/left + i put string here because of constant eror: 'int' object has no attribute 'strip' >> some cells already have just numbers?
        number_to_keep = '' # keeping numbers as a string, to convert later

        if 'min'in working_cell.lower(): # works for movies
            for char in working_cell:
                if char.isdigit():
                    number_to_keep += char

        elif 'season' in working_cell.lower(): # works for shows
            for char in working_cell:
                if char.isdigit():
                    number_to_keep += char

        
        if number_to_keep: # trying it only if found at least something
            row[col_index_local] = int(number_to_keep) # converting to int

        else:
            row[col_index_local] = working_cell # if there were no digits in the cell
````

````python
def average(data, column=9):
    '''
    This function calculates average value of given values in dataset.
    The function takes two arguments:
    1/ data
    2/ column (index) = by default set to 9, the current duration column

    Returns average
    '''

    # checking the quality of the data:
    if len(data) == 0:
        print('There is an empty dataset, nothing to explore.')


    sum_all = 0
    count_all = 0

    for row in data:
        working_cell = row[column]

        try:
            number = int(working_cell)
            sum_all += number
            count_all += 1
        
        except:
            continue

    
    average = sum_all/count_all

    return average

````

````python
def filter_data(row, filter_local):
    ''' 
    This function will take one row of the data at a time 
    and also the filter criteria in the form of a dictionary 
    where key is the column location and the value is the filter value expected.

    '''
    include_row_ll = True # By default, the row qualifies for the filtering criteria >> i didn't get it

    # (Task) Write a for loop that will loop over all the key value pairs of the filters and 
    # check the values in the row.

    for key_index, value_filter in filter_local.items(): # iterating over a dictionary of filters #5

        working_cell = row[key_index] # grab a cell
        #print('cell ', working_cell)


    # (Task) Inside for loop, write a if-else that will check if there are "," in that field.
    # (Task) If yes, then create a list out of it splitting it and then checking each value 
    # for the filter values.

        if "," in working_cell: # checks if there is a comma in a cell
            split_values = working_cell.split(",")   # splits the string in the cell by comma > makes a list
            cell_items_list = []

            for item in split_values: # iterates over the list of all the things in the cell
                cleaned_item = item.strip() # strips them of whitespaces (left/right)
                cell_items_list.append(cleaned_item) # adds to a list of items (splitted, cleaned)
                #print('list ', cell_items_list)
                
            if value_filter not in cell_items_list: # checks if the value_filter (e.g., country) is in the list of items
                include_row_ll = False
            
# (Task) Else, directly check it for the filter values.
        else:
            if working_cell.strip() != value_filter:
        # only happens if the cell didn't have any commas
                include_row_ll = False
        
    return include_row_ll
````

````python
def elements_count_mod_filters(data, col_index=1, sep_local=',', filters_local={}):
    '''
    The function counts unique elements for a particular column
    in each row and saves them in a dictionary by a unique key.

    It works even for cells that have more than one item (e.g., more than one
    genre is listed).

    It takes several arguments:
    1/ data -- to iterate over.
    2/ col_index -- stands for an index of a column we would like to analyse; by default takes second column
    3/ sep_local -- stands for a separator we want to use; the default separator is set to ',' .
    4/ filters_local -- stands for a dictionary of desired filters; default set to empty dictionary.

    Returns a dictionary of unique elements (keys) and their counts (values).
    '''


    unique_elements_filtered= {} # create empty dictionary to fill it with unique keys + counts

    # check, if data exists:
    #if len(data) == 0:
    #    print('There is an empty dataset, nothing to explore.')
    #    return {} # we need to return at least something

    # check if column exists:
    if col_index >= len(data[0]):
        print('Wrong index.')
        return {}



    for row in data: # we need to call the function we created to filter the rows multiple times
            # question is, how do we save output + count
            
            if filter_data(row, filters_local) == False: # checks if the row passes the filter; re-use the function
                continue

            else: 
                 

                working_cell = row[col_index] # which cell are we in?
    

                if sep_local in working_cell: # if there is a given separator in a cell, split
                    split_values = working_cell.split(sep_local)


                    for value in split_values:
                        element = value.strip()

                        if element not in unique_elements_filtered:
                            unique_elements_filtered[element] = 1

                        else: # doesn't work
                            unique_elements_filtered[element]+= 1

            
                else: # doesn't work
                    element = working_cell.strip()
                    if element not in unique_elements_filtered: # only works if there were no sep_loc in the cell
                        unique_elements_filtered[element] = 1

                    else: # doesn't work
                        unique_elements_filtered[element]+= 1
                 


    unique_elements_filtered_sorted = dict(sorted(unique_elements_filtered.items(),key=lambda item: item[1], reverse=True))

    return unique_elements_filtered_sorted
        



    
````
